## 3-7-1

```python
import numpy as np

arr = np.array([1, 2, 3,4, 5])

print(f"original: {arr}")
print(f"add: {arr + 1}")
print(f"sub: {arr - 1}")
print(f"mul: {arr * 2}")
```

## 3-7-2

```python
import numpy as np

arr = np.array(range(12))
arr = arr + 1
arr = arr.reshape((3, 4))

print(f"original: {arr}")
print(f"add: {arr + 1}")
print(f"sub: {arr - 1}")
print(f"mul: {arr * 2}")
```

## 3-7-3

```python
import math
import numpy as np
import matplotlib.pyplot as plt


x = np.arange(1, 20, 1)
y = np.log(x)

plt.plot(x, y)
plt.title("log curve")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

## 3-7-4

```python
import string
import numpy as np
import matplotlib.pyplot as plt

x = list(string.ascii_uppercase[0:10])
y = np.random.randint(5, 10, (10))

plt.bar(x, y)
plt.title("Bar Graph")
plt.show()
```

## 3-7-5

```python
import numpy as np
import matplotlib.pyplot as plt

labels = ['A', 'B', 'C', 'D']
values = np.array([1, 2, 3, 4], dtype=int)

plt.pie(values, labels=labels)
plt.show()
```

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

arr = np.arange(1,13).reshape((3, 4))
scaler = MinMaxScaler()
data = scaler.fit_transform(arr)
print(data)
```

```python
import numpy as np
from sklearn.preprocessing import StandardScaler

arr = np.arange(1, 13).reshape((3, 4))
scaler = StandardScaler()
data = scaler.fit_transform(arr)
print(data)
```
